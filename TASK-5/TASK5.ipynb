```python
#HISTOGRAM FOR AGE
plt.figure(figsize=(8,5))
plt.hist(t1["Age"],bins=20)
plt.title("AGE DISTRUBUTION")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()

#OBSERVATION
#Most passengers were between 20 and 40 years old. There were fewer children and elderly passengers. The age distribution is slightly right-skewed.
```

```python
#histogram for Fare
plt.figure(figsize=(8,5))
sns.histplot(t1["Fare"],bins=20,kde=True)
plt.title("FARE DISTRUBUTION")
plt.xlabel("Fare")
plt.ylabel("Count")
plt.show()

#OBSERVATION
#Most passengers paid lower fares.
#Only a small number of passengers paid very high fares.
#The fare distribution is positively (right) skewed due to high-fare passengers.
```

```python
# BOXPLOT
plt.figure(figsize=(6,4))
sns.boxplot(x=t1["Age"])
plt.title("BOXPLOT FOR AGE")
plt.show()

#OBSERVATION
#he median age is approximately 28 years. Half of the passengers are between about 22 and 34 years old. A few elderly passengers appear as outliers.
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=t1["Fare"])
plt.title("BOXPLOT FOR FARE")
plt.show()

#OBSERVAIONS
#Most passengers paid relatively low fares.Several high-fare passengers appear as outliers.The fare values are widely spread, indicating variation in ticket prices.
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x="Age",y="Fare",hue="Survived",data=t1)
plt.title("SCATTERPLOT DISTRIBUTION")
plt.show()

#OBSERVATIONS
 #The scatter plot shows the relationship between passengers' Age and Fare.
# There is no strong linear relationship between Age and Fare, as passengers of all ages paid different ticket fares.
# Most passengers paid lower fares regardless of their age.
# A few passengers paid very high fares, which appear as outliers.
# The different colors (Survived = 0 and 1) indicate that passengers who paid higher fares were generally more likely to survive.
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x="Age",y="Fare",hue="Sex",data=t1)
plt.title("scatterplot for AGE VS FARE")
plt.show()
# Observation:
# Age and Fare do not show a strong relationship.
# Most passengers paid lower fares.
# A few passengers paid very high fares.
# Both male and female passengers are present across different age groups and fare ranges.
```

```python
sns.pairplot(
    t1[["Age","Fare","Pclass","Survived"]])
plt.show()

# Observation:
# The pairplot compares the relationships between Age, Fare, Passenger Class, and Survival.
# First-class passengers generally paid higher fares.
# Passengers with higher fares appear to have a higher survival rate.
# Age does not clearly separate survivors from non-survivors.
# Passenger Class and Fare show a stronger relationship with Survival than Age.
```

```python
#COORELATION HEATMAP
temp = t1.copy()

temp["Sex"] = temp["Sex"].map({"male":0,"female":1})

corr = temp[["Survived","Pclass","Age","Fare","SibSp","Parch","Sex"]].corr()

plt.figure(figsize=(8,6))

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()


# Observation:
# Pclass has a negative correlation with Survival.
# Fare has a positive correlation with Survival.
# Sex shows a positive correlation with Survival (females had higher survival).
# Age has a weak correlation with Survival.
# Overall, Pclass, Fare, and Sex are the strongest factors related to Survival.
```

```python
#RELATIONSHIPS AND TRENDS

sns.countplot(x="Sex", hue="Survived", data=t1)
plt.title("Survival by Gender")
plt.show()

# Observation:
# Female passengers had a higher survival rate than male passengers.
# More male passengers did not survive.
# Gender appears to have influenced survival.
```

```python
sns.countplot(x="Pclass", hue="Survived", data=t1)
plt.title("Survival by Passenger Class")
plt.show()

# Observation:
# First-class passengers had the highest survival rate.
# Third-class passengers had the highest number of deaths.
# Passenger Class is strongly related to survival.
```
